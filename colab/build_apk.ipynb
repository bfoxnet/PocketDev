# PocketDev - build the APK on Google Colab

1. Upload `pocketdev.zip` (created by `make_zip.sh`) to this notebook's file browser.
2. Run the cells in order. The first build downloads Gradle + all dependencies (~5-10 min).
3. Download `app/build/outputs/apk/debug/app-debug.apk` when done.

```python
%%bash
# 1. Java 17 + Android SDK cmdline tools
set -e
apt-get -qq update
apt-get -qq install -y openjdk-17-jdk-headless >/dev/null
export JAVA_HOME=/usr/lib/jvm/java-17-openjdk-amd64
export PATH="$JAVA_HOME/bin:$PATH"
java -version 2>&1

if [ ! -d /opt/android-sdk/cmdline-tools/latest ]; then
  mkdir -p /opt/android-sdk/cmdline-tools
  curl -sSL -o /tmp/tools.zip \
    https://dl.google.com/android/repository/commandlinetools-linux-11076708_latest.zip
  unzip -q /tmp/tools.zip -d /opt/android-sdk/cmdline-tools
  mv /opt/android-sdk/cmdline-tools/cmdline-tools /opt/android-sdk/cmdline-tools/latest
fi
yes | /opt/android-sdk/cmdline-tools/latest/bin/sdkmanager --licenses >/dev/null 2>&1 || true
echo 'SDK ready.'
```

```python
%%bash
# 2. Unzip the uploaded project (skip if already extracted)
set -e
if [ ! -d /content/pocketdev ]; then
  mkdir -p /content/pocketdev
  unzip -q /content/pocketdev.zip -d /content/pocketdev
fi
ls /content/pocketdev
```

```python
%%bash
# 3. Build (first run downloads Gradle 8.10.2 + deps: 5-10 min)
set -e
export JAVA_HOME=/usr/lib/jvm/java-17-openjdk-amd64
export ANDROID_HOME=/opt/android-sdk
cd /content/pocketdev
chmod +x gradlew
./gradlew --no-daemon assembleDebug
```

```python
%%bash
# 4. Locate the APK
ls -lh /content/pocketdev/app/build/outputs/apk/debug/
```
